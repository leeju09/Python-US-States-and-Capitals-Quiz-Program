```python
# Import

import random

# ===========================================================================
# Part 1: Data Structure Setup
# ===========================================================================

# Step 1: Create the states_capitals dictionary with all 50 U.S. states
states_capitals = {
    "Alabama": "Montgomery",
    "Alaska": "Juneau",
    "Arizona": "Phoenix",
    "Arkansas": "Little Rock",
    "California": "Sacramento",
    "Colorado": "Denver",
    "Connecticut": "Hartford",
    "Delaware": "Dover",
    "Florida": "Tallahassee",
    "Georgia": "Atlanta",
    "Hawaii": "Honolulu",
    "Idaho": "Boise",
    "Illinois": "Springfield",
    "Indiana": "Indianapolis",
    "Iowa": "Des Moines",
    "Kansas": "Topeka",
    "Kentucky": "Frankfort",
    "Louisiana": "Baton Rouge",
    "Maine": "Augusta",
    "Maryland": "Annapolis",
    "Massachusetts": "Boston",
    "Michigan": "Lansing",
    "Minnesota": "St. Paul",
    "Mississippi": "Jackson",
    "Missouri": "Jefferson City",
    "Montana": "Helena",
    "Nebraska": "Lincoln",
    "Nevada": "Carson City",
    "New Hampshire": "Concord",
    "New Jersey": "Trenton",
    "New Mexico": "Santa Fe",
    "New York": "Albany",
    "North Carolina": "Raleigh",
    "North Dakota": "Bismarck",
    "Ohio": "Columbus",
    "Oklahoma": "Oklahoma City",
    "Oregon": "Salem",
    "Pennsylvania": "Harrisburg",
    "Rhode Island": "Providence",
    "South Carolina": "Columbia",
    "South Dakota": "Pierre",
    "Tennessee": "Nashville",
    "Texas": "Austin",
    "Utah": "Salt Lake City",
    "Vermont": "Montpelier",
    "Virginia": "Richmond",
    "Washington": "Olympia",
    "West Virginia": "Charleston",
    "Wisconsin": "Madison",
    "Wyoming": "Cheyenne"

}
```

```python
# Step 2: Data Validation
print("=" * 50)
print("DATA VALIDATION")
print("=" * 50)

# Print total number of states in dictionary
print(f"Total of states in dictionary: {len(states_capitals)}")

# Print the first 5 states and capitals alphabetically
print("\nFirst 5 states and capitals (alphabetically):")
sorted_states = sorted(states_capitals.keys())
for state in sorted_states[:5]:
  print (f" {state}: {states_capitals[state]}")

# Check that we have exactly 50 state-capital pairs
if len(states_capitals) == 50:
  print("\n Validation Passed: Exactly 50 state-capital pairs found.")
else:
  print("\n Validation Failed: Expected 50 pairs, but found {len(states_capitals)}.")
```

```python
# =========================================================================
# Part 2: Quiz implementation
# =========================================================================

# Step 1: Random Question Selection
def select_quiz_states (num_questions=5):
  """
    Randomly select a given number of unique states from the dictionary.
    Uses random.sample() to guarantee no state is repeated in one session.
  """
  return random.sample(list(states_capitals.keys()), num_questions)

# Step 2: Answer Checking (case-insensitive)
def check_answer_basic(user_answer, correct_capital):
  """
    Case-insensitive comparison between user input and correct capital.
    Example: 'sacramento' matches 'Sacramento'.
  """
  return user_answer.strip().lower() == correct_capital.strip().lower()

# Step 2 and 3: Main Quiz Loop with Score Tracking
def run_core_quiz():
  """
    Run the quiz using only Part 2 requirements:
      - Randomly select 5 unique states
      - Ask each question with a running number [1], [2], ...
      - Provide immediate Correct! / Incorrect feedback
      - Track correct count and display final score
  """
  selected_states = select_quiz_states(5)
  correct_count = 0

  for i, state in enumerate(selected_states, start=1):
    correct_capital = states_capitals[state]

    # Display question with running question number
    print(f"[{i}] What is the capital of {state}?")
    user_answer = input("Your answer: ").strip()

    # Provide immediate feedback
    if check_answer_basic(user_answer, correct_capital):
      print("Correct!")
      correct_count += 1
    else:
      print(f"Incorrect. The correct answer is {correct_capital}")
    print()

  # Display final score
  print(f"Your Score is {correct_count} out of 5.")
```

```python
# =========================================================================
# Part 3: Program Enhancement
# =========================================================================

# Step 1: Input Validation

def get_valid_answer (prompt= "Your answer: "):
  """
   Keep prompting until a non-empty answer is given.
    Strip leading/trailing whitespace automatically.
  """
  while True:
    answer = input(prompt).strip()
    if answer:
      return answer
    print("Please enter an answer (it cannot be empty).")


# Step 1: Enhanced spelling-variation matching

def is_answer_correct(user_answer, correct_capital):
  """
   Enhanced comparison that handles:
     - Case insensitivity ('sacramento' == 'Sacramento')
     - Punctuation ('St. Paul' == 'St Paul')
     - 'Saint' vs 'St' variations
  """

  # Normalize: lowercase, strip, remove periods

  user_norm = user_answer.lower().strip().replace(".", "")
  correct_norm = correct_capital.lower().strip().replace(".", "")

  # Collapse multiple spaces into one

  user_norm = " ".join(user_norm.split())
  correct_norm = " ".join(correct_norm.split())

  # Direct match

  if user_norm == correct_norm:
    return True

  # Handle "saint" vs "st" variations

  user_variant = user_norm.replace("saint", "st")
  correct_variant = correct_norm.replace("saint", "st")
  return user_variant == correct_variant

# Step 2: Welcome Message
def display_welcome():
  """ Display a welcome banner explaining the quiz."""
  print("="* 50)
  print("Welcome to the United States, States and Capitals Quiz!")
  print("You will be asked about 5 randomly selected states.")
  print("="* 50)
  print()

# Step 2: Encouraging Feedback

def get_encouragement(score, total):
  """ Return an encouraging message based on the final score. """
  if score == total:
    return "Perfect! Excellent knowledge of the U.S."
  elif score == 4:
    return "Great job! You know your state capitals well!"
  elif score == 3:
    return "Good work! You got more than half correct!"
  elif score >= 1:
    return "Keep studying! You'll improve with practice!"
  else:
    return "Don't give up! Review the state capitals and try again!"


# Full Enhanced Quiz
def run_enhanced_quiz():
  """
  Run the complete quiz combining Part 2 mechanics with Part 3 enhancements.
  """

  display_welcome()

  selected_states = select_quiz_states(5)
  correct_count = 0
  total_questions = len(selected_states)

  for i, state in enumerate(selected_states, start= 1):
    correct_capital = states_capitals[state]

    # Show progress: question X of 5
    print(f"[{i}] What is the capital of {state}? (Question {i} of {total_questions})")
    user_answer = get_valid_answer("Your answer: ")

    if is_answer_correct(user_answer, correct_capital):
      print("Correct!")
      correct_count += 1
    else:
      print(f"Incorrect. The answer is {correct_capital}.")
    print()

  # Final Score and encouragement
  print("="* 50)
  print("Quiz Complete!")
  print(f"Your score is {correct_count} out of {total_questions}.")
  print(get_encouragement(correct_count, total_questions))
  print("="* 50)
```
